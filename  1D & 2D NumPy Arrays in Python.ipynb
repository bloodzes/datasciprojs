```python
import numpy as np
# assign 1D array of numbers to var a
a = np.array([10,15,16,20,25,28])
d = np.array([1,2,3,4,5,6,])
print (a)
```

```python
#2D arrays are like a full spreadsheet table, list of lists
b = np.array([[2,5,6],[3,3,12]])
c = np.array ([[20,10,10],[15,20,10]])
print (b)
print (c)
```

```python
# in np : means everything
print (b[0, :])
```

```python
z = b + c
print (z)
```

```python
y = b * c
print (y)
```

```python
# dot product below multiplies matching elements, then adds everything up
# NOTE: must be same shape, i.e. 6 elements, x 6 elements for 1D array.
np.dot(a,d)
```

For 2D arrays, axis is used 

b.sum(axis=0)   #  sum down each column
b.sum(axis=1)   #   sum across each row

Memory trick: axis=0 collapses rows (goes down), axis=1 collapses columns (goes across).

```python
b.sum(axis=0)
```

Quick Cheat Sheet
Task	Code
- Make 1D array| 	np.array([1,2,3])
- Make 2D array|	np.array([[1,2],[3,4]])
- Check size/shape|	.shape, .ndim, .size
- Pick element|	a[2] or b[row, col]
- Slice|	a[1:4], b[:, 0]
- Math on all elements|	a * 2, a + b
- Summaries|	.sum(), .mean(), .max()
- Row/column totals	| axis=1 / axis=0

```python
#define strings first

str1 = 'Lorem ipsum dolor! diam amet, consetetur Lorem magna. sed diam nonumy eirmod tempor. diam et labore? et diam magna. et diam amet.'

class TextAnalyzer (object):

    def __init__ (self, text):
        formattedtext= text.replace(',','').replace('!','').replace('.','').replace('?','')
        
    #make text lowercase
        formattedtext = formattedtext.lower()
    
    #stores formattedtext into instance attribute so that other methods inside the TextAnalyzer class can access and use
        self.fmtText = formattedtext
    
    def freqAll(self):
        
        wordList = self.fmtText.split()
        # create dictionary
        FreqMap= {}
        #convert to set to remove duplicates
        for word in set(wordList): 
            FreqMap[word]= wordList.count(word)
        return FreqMap
        
        
    def freqOf(self,word):
        # get frequency map
        
        freqDict = self.freqAll()
        #ensure word we are looking up is lowercase
        word = word.lower()
        if word in freqDict:
            return freqDict[word]
        else:
            return 0
        
        
analyzed = TextAnalyzer(str1)
    
    #calling function 
print("Formatted Text:", analyzed.fmtText)
    
word = 'lorem'
frequency = analyzed.freqOf(word)
print("Word",word, "appears", frequency, "times")
        
    
    
```

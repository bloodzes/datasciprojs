Sets are a type of unordered collection, meaning that they do not record **element position** like List or Tuples.

Like Dictionaries, they are denoted with curly brackets but are separated by **commas instead of colons**


```python
Set1= {"pop", "rock", "indie", "blues", "soul", "pop"}
```

```python
#we can also convert a list into a set, sets automatically remove duplicates.
example_list = {'hey', 'im', 'in', 'love','with', 'love'}
example_set =set(example_list)
print (example_set)
```

```python
# to add or remove elements from a set
# example_list.add() or example_list.remove() can be used
# to check if elements are in a set, use the code 'in'
'love' in example_list
```

```python
#lastly, it is helpful to think of individual sets as venn diagrams
#two sets can overlap and its common elements can be established using &
#for example
example_set2 = {'hey', 'im', 'not', 'love'}
example_set3 = example_set & example_set2
print (example_set3)
```

```python
#to combine all unique elements from two sets, we use .union()
#example
example_set.union(example_set2)
```

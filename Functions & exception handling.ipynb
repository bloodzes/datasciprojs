```python
#sorted vs sort
#sorted creates a new list whereas sort doesn't create a new list
# For example

album_ratings = [1,2,8,5,4,3,4.5,3,3]
sorted_album_ratings = sorted (album_ratings) # new list sorted_album_ratings
print (sorted_album_ratings)

# sort(), sorts the list without creating a new list
album_ratings.sort()
print (album_ratings)
```

An exception is an error that occurs during code execution, when raised, it will halt the execution of the code if is not prepared to handle it.
- exceptions can be categorised into, ArithmeticError, IOError, ValueError, etc
- Refer to the link for a list of them (https://docs.python.org/3/library/exceptions.html)

```python
# ZeroDivisionError example
try:
    1/0
except Exception as e:
    print(type(e).__name__, "-", e)
```

```python
a = 1

try:
    b = int(input("Please enter a number to divide a"))
    a = a/b
except ZeroDivisionError:
    print("The number you provided cant divide 1 because it is 0")
except ValueError:
    print("You did not provide a number")
except:
    print("Something went wrong")
else:
    print("success a=",a)
finally:
    print("Processing Complete")
```

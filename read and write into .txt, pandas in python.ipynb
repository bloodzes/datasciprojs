```python
# open() creates a file object and accesses data within text file

file = open('file.txt', 'r') #'r'for reading, 'w'for writing 'a' for appending
```

```python
#adding a 'with' statement automatically closes the file when operations within indented block
#are completed,considered best practice.
# Open the file using 'with' in read ('r') mode
with open('file.txt', 'r') as file:

    #read method to read entire content
    file_content = file.read()
    
    print(file_content)
```

```python
#python also allows reading of one line at a time, and storing each line as 
#individual elements in a list

with open('file.txt', 'r') as file:
    line1 = file.readline()
    line2 = file.readline()
    
    #OR you could read all lines and save as list
    
with open('file.txt','r') as file:
    FileasList = file.readlines()

    #ecetera
    #you can perform specific actions at each line, for example
    print (line1.strip()) #strip() removes trailing characters in each new line
    if 'important' in line2:
        print ('this is important!')
        
#typically, loop functions are used to read line by line
while True:
    line = file.readline()
    if not line:
        break #stops when no more lines to read
    print(line)
```

```python
# moves file pointer to particular position in bytes
file.seek(10) #moves to the 11th byte (0-based index)
characters = file.read(5) #reads the next 5 characters
```

### Writing files with Open

```python
examplewrite = '/example1.txt'
with open(examplewrite, 'w') as writefile:
    writefile.write('this is line a')


#Possible to write multiple lines as well

with open(examplewrite, 'w') as writefile:
    writefile.write('this is line A\n')
    writefile.write('This is line B\n')
    
##Note that setting the mode to 'w' overwrites all existing data in the file
```

### Appending files allow you to write to files without losing any existing data

```python
# Write a new line to text file

with open('/Example2.txt', 'a') as testwritefile:
    testwritefile.write("This is line C\n")
    testwritefile.write("This is line D\n")
    testwritefile.write("This is line E\n")
    
    # Verify if the new line is in the text file

with open('/Example2.txt', 'r') as testwritefile:
    print(testwritefile.read())
```

### Copying files 

```python
# Copy file to another

with open('/Example2.txt','r') as readfile:
    with open('/Example3.txt','w') as writefile:
          for line in readfile:
                writefile.write(line)
```

### Loading data using Pandas Library for data analysis
- For structured data, designed to handle data in various formats
    - tabular data, time series data and more.
    
Data Structures: Pandas offers two primary data structures

- DataFrame and Series.

    -A DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns). Think CSV files, excel spreadsheets, SQL databases and more.

    -A Series is a one-dimensional labeled array, essentially a single column or row of data.

Data Import and Export: 
- Pandas makes it easy to read data from various sources, including CSV files, Excel spreadsheets, SQL databases, and more. It can also export data to these formats, enabling seamless data exchange.

Data Merging and Joining: 
- You can combine multiple DataFrames using methods like merge and join, similar to SQL operations, to create more complex datasets from different sources.

Efficient Indexing: 
- Pandas provides efficient indexing and selection methods, allowing you to access specific rows and columns of data quickly.

```python
import pandas as pd
csv_path = 'file1.csv'
df = pd.read_csv(csv_path)

#create a series name s with numeric data
data = [10, 20, 30, 40, 50]
s = pd.Series(data) 
print(s)
```

Other essential attributes and methods to manipulate and analyze data:

- values: Returns the Series data as a NumPy array.

- index: Returns the index (labels) of the Series.

- shape: Returns a tuple representing the dimensions of the Series.

- size: Returns the number of elements in the Series.

- mean(), sum(), min(), max(): Calculate summary statistics of the data.

- unique(), nunique(): Get unique values or the number of unique values.

- sort_values(), sort_index(): Sort the Series by values or index labels.

- isnull(), notnull(): Check for missing (NaN) or non-missing values.

- apply(): Apply a custom function to each element of the Series.

```python
# create dataframe from dictionary
import pandas as pd
data1 = {'Name': ['Alice', 'bob', 'charlie', 'david'], 
              'Age': [20,23,24,25],
              'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago'] }

df = pd.DataFrame(data1)

print(df)
print(df['Name'])
```

```python
# displaying single column
print(df['Name']) 
```

```python
#displaying data at index 0-2, and columns 0-1
print (df.iloc[0:3,0:2]) 
```

```python
# display data by label (rows 0:3, city header)
print (df.loc[0:3, 'City']) 
```

```python
df.to_csv('friendslist.csv', index=False)
```

DataFrames provide numerous attributes and methods for data manipulation and analysis, including:

- shape: Returns the dimensions (number of rows and columns) of the DataFrame.

- info(): Provides a summary of the DataFrame, including data types and non-null counts.

- describe(): Generates summary statistics for numerical columns.

- head(), tail(): Displays the first or last n rows of the DataFrame.

- mean(), sum(), min(), max(): Calculate summary statistics for columns.

- sort_values(): Sort the DataFrame by one or more columns.

- groupby(): Group data based on specific columns for aggregation.

- fillna(), drop(), rename(): Handle missing values, drop columns, or rename columns.

- apply(): Apply a function to each element, row, or column of the DataFrame.

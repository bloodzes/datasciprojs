Fundamental operations such as Else, and else if allows you to set branches of code that runs of the criteria is True

```python
# for example 
age = 18

if (age >18):
    
    print ('you can drink alcohol')
    
else:
    
    print ('move on')
    
```

elif statements allows multiple conditions 

```python
age = 18

if (age >18):
    
    print ('you can drink alcohol')
    
elif(age==18):

    print ('wait till you turn 19')
    
else:
    
    print ('move on')
```

Logic operators AND, OR function such that
- For OR, when either statement is TRUE, the **output will be TRUE, or 1.**
- For AND, both statements have to be **TRUE** to have an output of 1, otherwise, if one is false, **output will be false, or 0**

```python
#For example

album1 = 1970

if (album1 >=1970) and (album1 <1990):

    print ('the album was created in the 70s or 80s')

else :

    print ('the album was created in the 60s')
```

### Loops are mainly used on lists but can be used on Tuples as well. Loops help complete repeat tasks quickly and accurately


```python
#for loops
squares = ['red', 'yellow', 'green', ' purple']
for i in range(0,4):
    squares[i]= 'white'
    print (squares)
```

```python
# if rating of playlistratings is less than 6, exit the loop, using a while loop
PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]
i = 0
while i < len(PlayListRatings) and PlayListRatings[i] >= 6: #len is used here so that it prevents i from going past the list, and avoids a crash
        print(PlayListRatings[i])
        i += 1
```

```python
#while loops stops a process when a diff variable is encountered
#example
count = 1
while count <=20:
    print (count)
    count +=1

```

```python
# to print out elements between -5 and 5
for i in range (-5,6):
    print (i)
```

```python
Genres = ['rock', 'R&B', 'Soundtrack', 'R&B', 'Soul', 'pop']
for Genre in Genres:
    print (Genre)
```

note on the above

With a for loop, you can:

- Modify items: You can capitalize them (Genre.upper()) as they print.
- Filter items: You can use an if statement inside the loop to skip certain genres (e.g., only print if the genre is 'rock').
- Save to a database: You can send each individual genre name to a database one by one.

```python
#prints number from 1 to 15, but skips multiples of 3, stops when >12
for i in range(1, 16):
    if i % 3 == 0:
        continue  # skip multiples of 3
    if i > 12:
        break     # stop if number > 12
    print(i)
```

### So how do you know when to use each?

For Loops: Use for loops when you know the number of iterations in advance and want to process each element in a sequence. They are best suited for iterating over collections and sequences where the length is known.

While Loops: Use while loops when you need to perform a task repeatedly as long as a certain condition holds true. While loops are particularly useful for situations where the number of iterations is uncertain or where you're waiting for a specific condition to be met.

```python
# Tuples allow the combination of various data types into one

Tuple1 = (3.142,'pie','chart')

#Tuples can also be concatenated to form a new Tuple, Tuple2 in this case

Tuple2 = Tuple1+ (69,5)
print (Tuple2)
```

```python
#we can search for individual characters via indexing tuples

Tuple2[-1]
```

```python
# Now we slice the concatenated Tuple2 from characters 0 to 3

Tuple2[0:4]
```

```python
#for larger Tuples with multiple data types, we can run the function len() to find the length

len((Tuple2))
```

```python
#Since Tuples are Immutable objects, the data in the variables can't be amended.
# attempting to change it results in a TypeError 'Tuple'object does not support assignment
Ratings = (15,3,8,5,10,8,9,6,2,20)
print(Ratings)
Ratings[2]=3
```

```python
# However, we can assign a different Tuple to a new variable 'Rating'

Rating = ('order', 5, 'burgers', '&', 3, 'medium', 'fries')

# We can then sort it in ascending order using sorted() and 
# print the output

Ratingsascending = sorted(Ratings)
print(Ratingsascending)
```

```python
# Next, we have Lists, which are also ordered sequences of elements, but
#differ with tuples in that they are mutable

List1 = ['Marcus', 'Michael', 2, 3*6]
List1[0]= "Hey"  #since they are mutable, you can change the elements if needed
#Lists can be indexed too
print (List1[0:3])
print (List1)
```

```python
#to clone a list
List2 = List1[:]
print (List2)
```

```python
#to add an element to a list
List1.append ('9')
print (List1)
```

```python
#Note that the code to add element 5 and str '9'was executed multiple times
#the code below is used to delete elements
del List1 [4:8]
print (List1)
```

```python
#extending lists
List1.extend(List2)
print (List1)
```

```python
#to insert an element at a specific index, we use .insert(index,element)
List1.insert(4, 'whats up')
print (List1)
```

```python
List1[-1] = 50
print (List1)
```

```python
List3 = [10,67,39,29,49]
List1.extend(List3)
print (List1)
```
